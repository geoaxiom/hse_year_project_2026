```python
# Загрузка файла с демографией субьектов и клиническими данныим (Demographic and Clinical Data)
file_path = '/content/drive/MyDrive/mri/oasis_cross-sectional-5708aa0a98d82080.xlsx'
df_oasis_cross_sectional = pd.read_excel(file_path)
df_oasis_cross_sectional.head()
```

```python
df_oasis_cross_sectional.shape
```

```python
# Загрузка oasis freesurfer плоского набора из Parquet-файла в DataFrame
file_path = '/content/drive/MyDrive/mri/oasis_freesurfer_flat.csv'
df_oasis_freesurfer = pd.read_csv(file_path)
df_oasis_freesurfer.head()
```

```python
df_merged = pd.merge(
    df_oasis_freesurfer,
    df_oasis_cross_sectional,
    left_on='subject_id',
    right_on='ID',
    how='outer',
    indicator=True
)

print(f"Размерность df_oasis_freesurfer: {df_oasis_freesurfer.shape}")
print(f"Размерность df_oasis_cross_sectional: {df_oasis_cross_sectional.shape}")
print(f"Размерность merged dataframe: {df_merged.shape}")
```

```python
display(df_merged.head())
```

```python
# Проверим соединение наборов данных через колонку-метку (indicator=True)
# Находим строки, которые были ТОЛЬКО в левом DataFrame (нет совпадений в правом)
left_misses = df_merged[df_merged['_merge'] == 'left_only']

left_misses.shape
```

```python
# Находим строки, которые были ТОЛЬКО в правом DataFrame (нет совпадений в левом)
right_misses = df_merged[df_merged['_merge'] == 'right_only']

right_misses.shape
```

```python
# Находим ВCЕ пропуски вместе (строки без пары)
all_misses = df_merged[df_merged['_merge'] != 'both']

all_misses.shape
```

Почему в строк 425 а не 436

В полной клинической таблице oasis_cross-sectional числится 436 МРТ-сессий, но в официальном архиве FreeSurfer для OASIS-1 у 11 субъектов реконструкция поверхности (recon-all) в 2007 году не прошла контроль качества (из-за сильных артефактов движения у пациентов с деменцией или сбоев топологической коррекции), поэтому для них папки отсутствуют в архиве.

```python
# Эти 11 записи которых нет в архивах мы удалим
display(right_misses)
```

```python
# сохраним для проверки
right_misses.to_csv('/content/drive/MyDrive/mri/right_misses.csv', index=False)
```

```python
df_merged = df_merged[~df_merged['ID'].isin(right_misses['ID'])]
```

```python
df_merged.shape
```

```python
# Удалим служебные колонку для обьединения и дублекат ID = subject_id
df_merged.drop(columns=['ID', '_merge'], axis=1, inplace=True)
```

```python
df_merged.shape
```

```python
# Сохраняем результат обьединения на Google Drive в формате parquet
import os
output_parquet = os.path.join('/content/drive/MyDrive/mri/', 'oasis_merged.parquet')
df_merged.to_parquet(output_parquet, engine='pyarrow', compression='snappy')
print(f"Файл сохранен по пути: {output_parquet}")
```

```python
# Сохраняем результат  обьединения на Google Drive в формате csv
output_csv = os.path.join('/content/drive/MyDrive/mri/', 'oasis_merged.csv')
df_merged.to_csv(output_csv)
print(f"Таблица сохранена в: {output_csv}")
```
